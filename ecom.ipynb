Task 5: Geographical Sales Analysis  

```python
import matplotlib.pyplot as plt

# Top 10 states by orders
top_10_orders = (
    df.groupby("ship-state")["Order ID"]
    .count()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 States by Orders:")
print(top_10_orders)


# Top 10 states by revenue
top_10_revenue = (
    df.groupby("ship-state")["Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 States by Revenue:")
print(top_10_revenue)


# Horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_revenue.index,
    top_10_revenue.values
)

plt.xlabel("Total Revenue")
plt.ylabel("State")
plt.title("Top 10 States by Revenue")

plt.gca().invert_yaxis()

plt.show()


# High orders but low revenue
state_analysis = df.groupby("ship-state").agg(
    Total_Orders=("Order ID", "count"),
    Total_Revenue=("Amount", "sum")
)

state_analysis["Average_Order_Value"] = (
    state_analysis["Total_Revenue"] /
    state_analysis["Total_Orders"]
)

print("\nLow-Value Order States:")
print(state_analysis.sort_values("Average_Order_Value").head(10))
```

Task 6: Order Status & Fulfilment Analysis    

```python
import pandas as pd
import matplotlib.pyplot as plt


# 1. Order Status Distribution

status_count = df["Status"].value_counts()

print("Order Status Distribution:")
print(status_count)


# 2. Pie Chart

fig, ax = plt.subplots(figsize=(9, 7))

wedges, texts, autotexts = ax.pie(
    status_count.values,
    autopct=lambda pct: f"{pct:.1f}%" if pct > 2 else "",
    pctdistance=0.8,
    startangle=90,
)

ax.legend(
    wedges,
    status_count.index,
    title="Order Status",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1),
)

ax.set_title("Order Status Distribution")
plt.tight_layout()
plt.show()


# 3. Fulfilment Type vs Order Status

fulfilment_status = pd.crosstab(
    df["Fulfilment"],
    df["Status"]
)

print("Fulfilment vs Order Status:")
print(fulfilment_status)


# 4. Grouped Bar Chart

fulfilment_status.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.xlabel("Fulfilment Type")
plt.ylabel("Number of Orders")
plt.title("Fulfilment Type vs Order Status")

plt.xticks(rotation=0)
plt.legend(title="Order Status")

plt.show()


# 5. Cancellation Rate

cancellation_rate = (
    df.groupby("Fulfilment")["Status"]
    .apply(lambda x: (x == "Cancelled").mean() * 100)
)

print("Cancellation Rate:")
print(cancellation_rate)

# 6. Cancellation by Category

cancelled_category = (
    df[df["Status"] == "Cancelled"].groupby("Category").size()
    .sort_values(ascending=False)
)

print("Cancellation by Category:")
print(cancelled_category)


# 7. Cancellation by State

cancelled_state = (
    df[df["Status"] == "Cancelled"]
    .groupby("ship-state")
    .size()
    .sort_values(ascending=False)
)

print("Cancellation by State:")
print(cancelled_state)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Make sure Date is datetime
df["Date"] = pd.to_datetime(df["Date"])

# 1. Monthly Sales
monthly_sales = df.groupby(
    df["Date"].dt.to_period("M")
)["Amount"].sum()

# 2. Category Revenue
category_revenue = (
    df.groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)
# 3. Top 10 States Revenue
state_revenue = (
    df.groupby("ship-state")["Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# 4. Order Status
status_count = df["Status"].value_counts()

# 5. Fulfilment vs Status
fulfilment_status = pd.crosstab(df["Fulfilment"],df["Status"])


# 6. Category Quantity
category_quantity = (df.groupby("Category")["Qty"].sum().sort_values(ascending=False))

# DASHBOARD

fig, axes = plt.subplots(3, 2, figsize=(16, 14))

fig.suptitle("E-Commerce Sales Analysis Dashboard",fontsize=20)


# Chart 1 - Monthly Sales Trend
axes[0, 0].plot(monthly_sales.index.astype(str),monthly_sales.values,marker="o")

axes[0, 0].set_title("Monthly Sales Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Total Sales")
axes[0, 0].tick_params(axis="x", rotation=45)
axes[0, 0].grid(True)


# Chart 2 - Revenue by Category
axes[0, 1].bar(
    category_revenue.index,
    category_revenue.values
)

axes[0, 1].set_title("Revenue by Category")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Revenue")
axes[0, 1].tick_params(axis="x", rotation=45)


# Chart 3 - Top 10 States
axes[1, 0].barh(
    state_revenue.index,
    state_revenue.values
)

axes[1, 0].set_title("Top 10 States by Revenue")
axes[1, 0].set_xlabel("Revenue")
axes[1, 0].set_ylabel("State")
axes[1, 0].invert_yaxis()


# Chart 4 - Order Status
axes[1, 1].pie(
    status_count.values,
    labels=status_count.index,
    autopct="%1.1f%%"
)

axes[1, 1].set_title("Order Status Distribution")


# Chart 5 - Fulfilment vs Status
fulfilment_status.plot(
    kind="bar",
    ax=axes[2, 0]
)

axes[2, 0].set_title("Fulfilment Type vs Order Status")
axes[2, 0].set_xlabel("Fulfilment Type")
axes[2, 0].set_ylabel("Number of Orders")
axes[2, 0].tick_params(axis="x", rotation=0)
axes[2, 0].legend(title="Status",fontsize=5)


# Chart 6 - Category Quantity
axes[2, 1].bar(
    category_quantity.index,
    category_quantity.values
)

axes[2, 1].set_title("Quantity Sold by Category")
axes[2, 1].set_xlabel("Category")
axes[2, 1].set_ylabel("Quantity")
axes[2, 1].tick_params(axis="x", rotation=45)


# Adjust spacing
plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
